## Тема 4. Парадоксы в теории вероятностей

### "На что спорим?" (парадокс трёх дверей(Монти-Холла))

Перед участником находятся три закрытые двери. За одной дверью находится автомобиль, а за двумя другими — козы.

Участник выбирает одну из трёх дверей. После этого ведущий, который знает, где находится автомобиль, открывает одну из двух оставшихся дверей, за которой находится коза.

Затем участнику предлагают изменить свой первоначальный выбор и выбрать другую закрытую дверь.

Возникает вопрос: выгоднее оставить первоначальный выбор или поменять его?

Если участник не меняет свой выбор, вероятность выигрыша составляет:

$$P(A)=\frac{1}{3}$$

Если участник меняет свой выбор, вероятность выигрыша составляет:

$$P(B)=\frac{2}{3}$$

Таким образом, изменение первоначального выбора увеличивает вероятность выигрыша в два раза.

### Моделирование парадокса

Смоделируем 100000 игр. В каждой игре случайным образом размещается автомобиль, участник выбирает дверь, а ведущий открывает одну из оставшихся дверей с козой.

Сравним количество выигрышей при сохранении первоначального выбора и при его изменении.

### Вывод

В результате моделирования видно, что при сохранении первоначального выбора вероятность выигрыша приближается к 1/3, а при изменении выбора — к 2/3.

Таким образом, моделирование подтверждает парадокс Монти Холла: вероятность выигрыша при смене первоначального выбора в два раза выше, чем при сохранении своего решения.

```python
import numpy as np

# Количество игр
N = 100000

# Генератор случайных чисел
rng = np.random.default_rng(42)

# Счётчики выигрышей
wins_stay = 0
wins_switch = 0

# Моделирование игр
for i in range(N):

    # Случайное расположение автомобиля
    car = rng.integers(0, 3)

    # Первоначальный выбор участника
    choice = rng.integers(0, 3)

    # Ведущий открывает дверь с козой
    available = [
        door for door in range(3)
        if door != choice and door != car
    ]

    opened = rng.choice(available)

    # Оставшаяся закрытая дверь
    switch_choice = next(
        door for door in range(3)
        if door != choice and door != opened
    )

    # Выигрыш без смены выбора
    if choice == car:
        wins_stay += 1

    # Выигрыш при смене выбора
    if switch_choice == car:
        wins_switch += 1

# Расчёт вероятностей
P_stay = wins_stay / N
P_switch = wins_switch / N

# Вывод результатов
print(f"Количество игр: {N}")

print(f"\nБез смены выбора:")
print(f"Количество выигрышей: {wins_stay}")
print(f"Вероятность выигрыша: {P_stay:.3f}")

print(f"\nПри смене выбора:")
print(f"Количество выигрышей: {wins_switch}")
print(f"Вероятность выигрыша: {P_switch:.3f}")

print("\nТеоретические вероятности:")
print(f"Без смены выбора: {1/3:.3f}")
print(f"При смене выбора: {2/3:.3f}")
```
