```python
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon, Rectangle
import numpy as np
```

Смоделирую ситуацию, в которой нужно попасть в треугольник, находящийся внутри квадрата

```python
def darts(size):
    side = 10.0
    A = np.array([0.0, 0.0])
    B = np.array([side, 0.0])
    C = np.array([side / 2.0, side])
    tr_ver = [A, B, C]
    fig, ax = plt.subplots(figsize=(6, 6))
    
    ax.add_patch(Rectangle((0, 0), side, side, edgecolor='black', facecolor='none', lw=1.5))
    ax.add_patch(Polygon(tr_ver, facecolor='red', alpha=0.5, edgecolor='darkred', lw=2))

    # Случайные броски дротиков
    X = np.random.uniform(0, side, size)
    Y = np.random.uniform(0, side, size)

    inside_mask = (Y >= 0) & (Y <= 2 * X) & (Y <= 2 * (side - X))
    inside_count = np.sum(inside_mask)
    prob_empirical = inside_count / size

    ax.scatter(X[inside_mask], Y[inside_mask], marker='x', c='green', s=25, label='Попадание')
    ax.scatter(X[~inside_mask], Y[~inside_mask], marker='x', c='orange', s=25, label='Промах')

    ax.set_xlim(-0.5, side + 0.5)
    ax.set_ylim(-0.5, side + 0.5)
    ax.set_aspect('equal')
    ax.legend(loc='upper right')
    
    plt.show()

    print(f"В треугольник из {size} дротиков попало: {inside_count}")

darts(1000)
```
