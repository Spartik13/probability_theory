В коробке находится 6 красных и 6 синих шариков. Требуется оценить вероятность случайно вынуть 3 красных и 3 синих шарика.

Пользуясь классическим определением, получим:

$$P(A)=\frac{C_6^3\cdot C_6^3}{C_{12}^6}=0.4329$$

Оценим вероятность вынуть 3 красных и 3 синих шарика, используя от 10 до 10000 испытаний.

Построим график зависимости частотной вероятности от количества проведённых испытаний.

### Вывод

В результате моделирования видно, что при увеличении количества испытаний частотная вероятность приближается к классической вероятности, равной 0.4329.

Таким образом, моделирование подтверждает связь частотного и классического определений вероятности.

```python
import numpy as np
import matplotlib.pyplot as plt
from random import sample
from math import comb

# Классическая вероятность
P = comb(6, 3) * comb(6, 3) / comb(12, 6)

print(f"Классическая вероятность: {P:.4f}")

# Функция случайного выбора 6 шариков
def random_combination():
    urn = ['R'] * 6 + ['B'] * 6
    return sample(urn, 6)

# Количество испытаний
N_values = np.arange(10, 10001, 50)
P_A_values = []

# Моделирование эксперимента
for N in N_values:
    N_A = 0

    for _ in range(N):
        combination = random_combination()

        if combination.count('R') == 3:
            N_A += 1

    P_A_values.append(N_A / N)

print(f"Частотная вероятность: {P_A_values[-1]:.4f}")

# Построение графика
plt.figure(figsize=(10, 5))

plt.plot(N_values, P_A_values, label='Частота')
plt.axhline(y=P, color='red', linestyle='--', label='Классическая вероятность')

plt.xlabel('Количество испытаний N')
plt.ylabel('P(A)')
plt.title('Связь частотной и классической вероятности')

plt.legend()
plt.grid()
plt.show()
```
