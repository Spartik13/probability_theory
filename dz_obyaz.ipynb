Всего 216 различных способов для бросания трех кубиков

а) А - случай, при котором выпало 11 очков

```python
all = list(iter.product(range(1, 7), repeat=3))
eleven_points = list(a for a in all if sum(a) == 11)
round(len(eleven_points) / len(all), 4)
```

б) B - случай, при котором выпало больше 10 очков

```python
all = list(iter.product(range(1, 7), repeat=3))
eleven_points = list(a for a in all if sum(a) > 10)
round(len(eleven_points) / len(all), 4)
```
