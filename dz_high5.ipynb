```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, geom, hypergeom
```

### 5.1. Биномиальное распределение

**Задача.**

Вероятность успешного выполнения одного задания равна 0.3. Студент выполняет 5 независимых заданий. Случайная величина $X$ — число успешно выполненных заданий.

Данная случайная величина имеет биномиальный закон распределения с параметрами $n=5$, $p=0.3$, где $q=1-p=0.7$.

Биномиальное распределение представляет собой закон распределения числа наступлений события в $n$ независимых испытаниях, в каждом из которых вероятность наступления события одинакова.

Формула распределения:

$$P(X=m)=C_n^m p^m q^{n-m}$$

где $m=0,1,\ldots,n$.

Математическое ожидание и дисперсия:

$$M(X)=np$$

$$D(X)=npq$$

Среднее квадратическое отклонение:

$$\sigma(X)=\sqrt{D(X)}$$

```python
# Биномиальное распределение
n = 5
p = 0.3

X = np.arange(0, n + 1)
dist = binom(n, p)

P = dist.pmf(X)

# Характеристики
M = dist.mean()
D = dist.var()
sigma = dist.std()
mode = X[np.argmax(P)]

print("Биномиальный закон распределения")
print("\nЗакон распределения:")

for x, prob in zip(X, P):
    print(f"P(X = {x}) = {prob:.4f}")

print(f"\nМатематическое ожидание: {M:.3f}")
print(f"Дисперсия: {D:.3f}")
print(f"Среднее квадратическое отклонение: {sigma:.3f}")
print(f"Мода: {mode}")

# Многоугольник распределения
plt.figure(figsize=(8, 5))
plt.plot(X, P, marker='o')
plt.title("Биномиальное распределение")
plt.xlabel("Значение X")
plt.ylabel("P(X)")
plt.xticks(X)
plt.grid()
plt.show()
```

### 5.2. Закон распределения Пуассона

**Задача.**

В среднем за час на сервер поступает 2.5 запроса с ошибками. Случайная величина $X$ — число запросов с ошибками за один час.

Используем закон распределения Пуассона с параметром $\lambda=2.5$.

Распределение Пуассона применяется при большом числе испытаний и малой вероятности наступления события.

Формула распределения:

$$P(X=m)=\frac{\lambda^m e^{-\lambda}}{m!}$$

где $m=0,1,2,\ldots$

Математическое ожидание и дисперсия совпадают и равны параметру распределения:

$$M(X)=D(X)=\lambda$$

Среднее квадратическое отклонение:

$$\sigma(X)=\sqrt{\lambda}$$

```python
# Распределение Пуассона
lam = 2.5

X = np.arange(0, 13)
dist = poisson(lam)

P = dist.pmf(X)

# Характеристики
M = dist.mean()
D = dist.var()
sigma = dist.std()
mode = X[np.argmax(P)]

print("Закон распределения Пуассона")
print("\nЗакон распределения:")

for x, prob in zip(X, P):
    print(f"P(X = {x}) = {prob:.4f}")
print(f"P(X > 12) = {dist.sf(12):.3f}")

print(f"\nМатематическое ожидание: {M:.3f}")
print(f"Дисперсия: {D:.3f}")
print(f"Среднее квадратическое отклонение: {sigma:.3f}")
print(f"Мода: {mode}")

# Многоугольник распределения
plt.figure(figsize=(8, 5))
plt.plot(X, P, marker='o')
plt.title("Распределение Пуассона")
plt.xlabel("Значение X")
plt.ylabel("P(X)")
plt.xticks(X)
plt.grid()
plt.show()
```

### 5.3. Геометрическое распределение

**Задача.**

Вероятность успешного выполнения одной попытки равна 0.4. Попытки выполняются независимо до первого положительного исхода. Случайная величина $X$ — число проведённых попыток до первого успеха.

Используем геометрическое распределение с параметрами $p=0.4$ и $q=1-p=0.6$.

Геометрическое распределение описывает число испытаний, проведённых по схеме Бернулли до первого положительного исхода.

Формула распределения:

$$P(X=m)=pq^{m-1}$$

где $m=1,2,\ldots$

Математическое ожидание:

$$M(X)=\frac{1}{p}$$

Дисперсия:

$$D(X)=\frac{q}{p^2}$$

Среднее квадратическое отклонение:

$$\sigma(X)=\sqrt{D(X)}$$

```python
# Геометрическое распределение
p = 0.4

X = np.arange(1, 13)
dist = geom(p)

P = dist.pmf(X)

# Характеристики
M = dist.mean()
D = dist.var()
sigma = dist.std()
mode = X[np.argmax(P)]

print("Геометрический закон распределения")
print("\nЗакон распределения:")

for x, prob in zip(X, P):
    print(f"P(X = {x}) = {prob:.4f}")
print(f"P(X > 12) = {dist.sf(12):.3f}")

print(f"\nМатематическое ожидание: {M:.3f}")
print(f"Дисперсия: {D:.3f}")
print(f"Среднее квадратическое отклонение: {sigma:.3f}")
print(f"Мода: {mode}")

# Многоугольник распределения
plt.figure(figsize=(8, 5))
plt.plot(X, P, marker='o')
plt.title("Геометрическое распределение")
plt.xlabel("Значение X")
plt.ylabel("P(X)")
plt.xticks(X)
plt.grid()
plt.show()
```

### 5.4. Гипергеометрическое распределение

**Задача.**

В партии находится 12 изделий, из которых 5 бракованных и 7 качественных. Случайным образом выбирают 4 изделия без возвращения. Случайная величина $X$ — число бракованных изделий среди выбранных.

Используем гипергеометрическое распределение с параметрами $n=4$, $M=5$, $N=12$.

Гипергеометрическое распределение описывает число объектов, обладающих заданным свойством, среди $n$ объектов, случайно извлечённых без возвращения из совокупности $N$ объектов, $M$ из которых обладают этим свойством.

Формула распределения:

$$P(X=m)=\frac{C_M^m C_{N-M}^{n-m}}{C_N^n}$$

где $m=0,1,\ldots,\min(n,M)$.

Математическое ожидание:

$$M(X)=n\frac{M}{N}$$

Дисперсия:

$$D(X)=n\frac{M}{N-1}
\left(1-\frac{M}{N}\right)
\left(1-\frac{n}{N}\right)$$

Среднее квадратическое отклонение:

$$\sigma(X)=\sqrt{D(X)}$$

```python
# Гипергеометрическое распределение
N = 12
K = 5
n = 4

X = np.arange(0, 5)
dist = hypergeom(N, K, n)

P = dist.pmf(X)

# Характеристики
M = dist.mean()
D = dist.var()
sigma = dist.std()
mode = X[np.argmax(P)]

print("Гипергеометрический закон распределения")
print("\nЗакон распределения:")

for x, prob in zip(X, P):
    print(f"P(X = {x}) = {prob:.4f}")

print(f"\nМатематическое ожидание: {M:.3f}")
print(f"Дисперсия: {D:.3f}")
print(f"Среднее квадратическое отклонение: {sigma:.3f}")
print(f"Мода: {mode}")

# Многоугольник распределения
plt.figure(figsize=(8, 5))
plt.plot(X, P, marker='o')
plt.title("Гипергеометрическое распределение")
plt.xlabel("Значение X")
plt.ylabel("P(X)")
plt.xticks(X)
plt.grid()
plt.show()
```
